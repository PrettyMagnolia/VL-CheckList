```python
from collections import defaultdict


res = defaultdict(list)
import os
import json
base_dir = '/home/yifei/code/VL-CheckList/output/open_clip/itc'
for file in os.listdir(base_dir):
    if file.endswith('.json'):
        with open(os.path.join(base_dir, file), 'r') as f:
            data = json.load(f)
            full_name = file.split('.')[0]
            acc = data['total_acc']
            num = data['number_of_data']
            if 'Object' in full_name:
                key = '_'.join(full_name.split('_')[:3])
            else:
                key = '_'.join(full_name.split('_')[:2])
            res[key].append((full_name, acc, num))

for key in sorted(res.keys()):
    value = res[key]
    sum_num = 0
    for v in value:
        sum_num += v[2]
    avg_acc = 0
    for v in value:
        avg_acc += v[1] * v[2] / sum_num
    print(key, round(avg_acc * 100, 2))
```
